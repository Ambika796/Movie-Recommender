```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = pd.read_csv("movies.csv")
ratings = pd.read_csv("ratings.csv")
```

```python
movies.head()
ratings.head()
```

```python
movie_matrix = ratings.pivot_table(index='userId', columns='movieId', values='rating')
```

```python
movie_matrix = movie_matrix.fillna(0)
```

```python
similarity = cosine_similarity(movie_matrix.T)  # transpose for movie-to-movie similarity
```

```python
def recommend_movie(movie_name, movies, similarity):
    movie_index = movies[movies['title'] == movie_name].index[0]
    sim_scores = list(enumerate(similarity[movie_index]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:6]  # top 5 similar movies
    movie_indices = [i[0] for i in sim_scores]
    return movies['title'].iloc[movie_indices].tolist()
```

```python
recommend_movie("Toy Story (1995)", movies, similarity)
```
